```python
#Q2
data='car_fuel_efficiency_2026.csv'
df=pd.read_csv(data)
print("length or count:",len(df))
df.head()
```

```python
#Q3
print(df.columns)
len(df.fuel_type)
df.fuel_type.unique()
```

```python
#Q4
print("Null columns:",df.columns[df.isnull().sum()>0])
df.isnull().sum()
```

```python
#Q5
df[df.origin=="Asia"].describe()
print('Max fuel efficiency of Asian cars:', df[df['origin'] == 'Asia'].fuel_efficiency_mpg.max())
```

```python
#Q6
df.horsepower.describe()
df_median=df.horsepower.median()
df_mode=df.horsepower.mode()[0]
print("Previous median and mode:", df_median,df_mode)
df['horsepower'] = df['horsepower'].fillna(df_mode)
print("New horsepower median:",df.horsepower.median())
```

```python
print(df[df['horsepower'].isnull()])
```

```python
#Q7
df.head()
df1=df[df['origin']=='Asia']
df1_sub=df1[['vehicle_weight', 'model_year']].head(7)
X=df1_sub.to_numpy()
XTX=X.T.dot(X)
XTX_inv=np.linalg.inv(XTX)
Y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
W=XTX_inv.dot(X.T).dot(Y) #multiplication of inverse, transpose, and Y
result=W.sum()
print(result)
```
